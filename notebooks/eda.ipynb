```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['figure.dpi'] = 100

RANDOM_STATE = 42
RAW_PATH = Path('/home/nikitos/MLSystem/data/raw/UCI_Credit_Card.csv')
```

```python
df = pd.read_csv(RAW_PATH)
df = df.drop(['ID'], axis=1)
print(df.shape)
df.head(10)
```

```python
df.info()
```

```python
df.isna().sum().sort_values(ascending=False)
```

```python
df.describe()
```

```python
payment_values = df['default.payment.next.month'].value_counts()
payment_pct = df['default.payment.next.month'].value_counts(normalize=True) * 100

print(payment_values)
print(f'Дисбаланс: {payment_pct[1]}% класса 1')


fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=df, x='default.payment.next.month', ax=ax[0], palette='Set2')
ax[0].set_title('Распределение Payment')

ax[1].pie(payment_values, labels=payment_values.index, autopct='%1.1f%%',
          colors=['#66b3ff', '#ff9999'], startangle=90)
ax[1].set_title('Доля Payment')
plt.tight_layout()
plt.savefig('../reports/figures/payment_distribution.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
pay_cols = ['PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']

fig, axes = plt.subplots(2, 3, figsize=(16,9))
for i, col in enumerate(pay_cols):
    ax = axes[i // 3, i % 3]
    sns.countplot(data=df, x=col, hue='default.payment.next.month', ax=ax, palette='Set2')
    ax.set_title(f'{col} vs Payment')
    ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.savefig('../reports/figures/PAY_N_vs_Payment.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
bill_cols = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']

fig, axes = plt.subplots(2, 3, figsize=(16,9))
for i, col in enumerate(bill_cols):
    ax = axes[i // 3, i % 3]
    sns.kdeplot(data=df, x=col, hue='default.payment.next.month', fill=True,
                ax=ax, palette='Set1', common_norm=False)
    ax.set_title(f'{col} vs Payment')
    ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.savefig('../reports/figures/BILL_AMTN_vs_Payment.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
payamt_cols = ['PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']

fig, axes = plt.subplots(2, 3, figsize=(16,9))
for i, col in enumerate(payamt_cols):
    ax = axes[i // 3, i % 3]
    sns.kdeplot(data=df, x=col, hue='default.payment.next.month', fill=True,
                ax=ax, palette='Set1', common_norm=False)
    ax.set_title(f'{col} vs Payment')
    ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.savefig('../reports/figures/PAY_AMTN_vs_Payment.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(20, 20))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Корреляции признаков')
plt.savefig('../reports/figures/correlation_matrix.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python


from statsmodels.stats.outliers_influence import variance_inflation_factor

X_vif = df.drop(columns=['default.payment.next.month']).dropna()
vif_data = pd.DataFrame({
    'feature': X_vif.columns,
    'VIF': [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
})
print(vif_data.sort_values('VIF', ascending=False))

```
